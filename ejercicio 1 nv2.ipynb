```python
pacients = [ 
    "Maria|45|tos, Fiebre Alta, dolor_pecho", 
    "Luis|33|dolor_cabeza,faTiga", 
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos", 
    "Jordi|52|fatiga , tos", 
    "Anna|29|fiebre alta,dolor_cabeza" 
]
```

```python
def pacientes_separados_i_normalizados(lista_pacientes):
    lista = []
    for paciente in pacients:
        lista.append(paciente.split("|"))

    for paciente in lista:
    #     paciente[2] = ', '.join(sintoma.strip().title()
    #                            for sintoma in paciente[2].split(","))
        paciente[2] = paciente[2].title()
        paciente[2] = paciente[2].replace("_", " ")
        sintomas = paciente[2].split(",")

        lista_sintomas = []

        for sintoma in sintomas:
            lista_sintomas.append(sintoma.strip().title())

        paciente[2] = ", ".join(lista_sintomas)

    return lista
```

```python
pacientes_separados_i_normalizados(pacients)
```
