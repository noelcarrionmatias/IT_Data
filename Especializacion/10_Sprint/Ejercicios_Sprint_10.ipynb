# Nivel 1: Exploración básica con una API de laboratorio

API: https://jsonplaceholder.typicode.com/

## 1- Consulta API pública JSONPlacerholder

Importar la librería

```python
import requests
from getpass import getpass
import pandas as pd
```

### Lista de publicaciones

```python
response_posts = requests.get('https://jsonplaceholder.typicode.com/posts')
api_posts = response_posts.json()
api_posts
```

```python
list_posts = []
for post in api_posts:
    list_posts.append(post['title'])
```

### Lista de usuarios

```python
response_users = requests.get('https://jsonplaceholder.typicode.com/users')
api_users = response_users.json()
api_users
```

```python
list_user = []
for user in api_users:
    list_user.append(user['username'])

```

### Lista de tareas

```python
response_tareas = requests.get("https://jsonplaceholder.typicode.com/todos")
api_tareas = response_tareas.json()
api_tareas
```

```python
list_tareas = []
for tarea in api_tareas:
    list_tareas.append(tarea['title'])

```

## 2 - Muestra en pantalla

### Cantidad de cada recurso (publicaciones, usuarios y tareas)

```python
print(f'Longitud de posts: {len(list_posts)}')
print(f'Longitud de usuarios: {len(list_user)}')
print(f'Longitud de tareas: {len(list_tareas)}')
```

### Código de estado de cada petición

```python
print(f'Código de estado de petición posts: {response_posts.status_code}')
print(f'Código de estado de petición usuarios: {response_users.status_code}')
print(f'Código de estado de petición tareas: {response_tareas.status_code}')
```

### 3 - Hacer una consulta inexistente para obtener código error 404 y mostrarlo en pantalla

```python
response_invent = requests.get('https://jsonplaceholder.typicode.com/errors')
response_invent.status_code
```

### 4 - Petición POST para crear una publicación ficticia. Incluye título, cuerpo de mensaje y userid

Creo una nueva publicación

```python
miobra = requests.post(url='https://jsonplaceholder.typicode.com/posts',
              json={'userId':9999999999,
                    'id': 123456789,
                    'title':'Mi obra inventada',
                    'body':"Lorem Ipsum is simply dummy text of the printing and typesetting industry. Lorem Ipsum has been the industry's standard dummy text ever since 1966,"})
```

Compruebo si ha funcionado (codigo 201 es created) y el contenido. Si ha fallado muestra el error

```python
if miobra.status_code == 201:
    print("Publicación creada correctamente")
    print(miobra.json())
    print(f'Código de estado: {miobra.status_code}')
else:
    print("Error:", miobra.status_code)
```

### 5 - Petición PATH para modificar parcialmente una publicación existente

```python
original = requests.get('https://jsonplaceholder.typicode.com/posts/1')
print(original.json())
modify = requests.patch('https://jsonplaceholder.typicode.com/posts/1', json={'userId':987654321})
print(modify.json())
```

### 6 - Petición DELETE para borrar una publicación 

```python
original = requests.get('https://jsonplaceholder.typicode.com/posts/12')
print(original.json())
delete = requests.delete('https://jsonplaceholder.typicode.com/posts/12')
print(delete.json())
```

# Nivel 3: API de Open Data Barcelona

URL Api base: https://opendata-ajuntament.barcelona.cat/data/api/ y se accede desde https://opendata-ajuntament.barcelona.cat/data/api/action/

### Resumen docu API OpenDataBCN:

#### API gestión de catálogo
- current_package_list_with_resources: catálogo de datasets y recursos
- package_search: búsqueda dentro del catálogo
- resource_show: devuelve metadatos del dataset a partir del id dataset
- datastore_search: devuelve información de cada recurso

#### API de consulta sobre recursos CSV
"API CKAN disponible" hay dos endpoints comunes para todos los CSV que permiten buscar dentro del recurso con parametros o consulta SQL:
- datastore_search: https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search
- datastore_search_sql: https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search_sql

#### Parámetros de entrada en datastore_search
- resource_id (string - obligatori) - id del recurs CSV sobre el que volem cercar
- filters (diccionari) - condicions que s’han de complir. Exemple: {“direccio”: “%carrer%”}
- q (string o diccionari) - cerca en full text. Es pot especificar un string per a cercar-lo a tots els camps o un diccionari per a cercar a camps concrets, p.e: {“key1”: “a”, “key2”: “b”}
- distinct (boolean) - retorna sols les files diferents
- plain (boolean) - tracta la consulta com si fos text pla
- language (string) - llenguatge de la consulta en full text
- limit (int) - número màxim de resultats. Per defecte retorna 100 resultats.
- offset (int) - número de resultats que s’han de saltar. Útil per fer una paginació de resultats.
- fields (string) - llistat de camps que s’inclouràn a la resposta. Per defecte retorna tots els camps amb el mateix ordre que al CSV
- sort (string) - noms dels camps per els que ordenar separats per comes: “nom, codi desc”

Ejemplo: buscar fuentes públicas para beber en la calle Mallorca:
https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search?resource_id=21f7a4df-2e73-45f8-8c6d-0b3db8c21527&limit=5&q={%22INVENTARI_CARRER%22:%22%Mallorca%%22}


## 1 - Utiliza la API para crear un dataset de tu interés con package_search i package_show

### current_package_list_with_resources

```python
list_recursos = requests.get("https://opendata-ajuntament.barcelona.cat/data/api/action/current_package_list_with_resources")
recursos_api_bcn = list_recursos.json()
```

```python
recursos_api_bcn
```

Obtengo una muestra (5) del nombre, nombre técnico e id del dataset, nombre e id del resource del dataset.

```python
datasets = recursos_api_bcn.get('result', [])

for data in datasets[:5]:
    print(f'Nombre del dataset: {data['title_translated']['es']}')
    print(f'Nombre del dataset técnico: {data['name']}')
    print(f'ID del dataset: {data['id']}')
    print(f'Nombre del resource: {data['resources'][0]['name']}')
    print(f'ID del resource del dataset: {data['resources'][0]['id']}\n')
```

### package_search

Busco con package_search un paquete a partir del parámetro 'q' y con el id '6aa3416d-ce1a-494d-861b-7bd07f069600'.

```python
url_search = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"
params_search = {
    'q': f'id:1412100d-b0cd-4894-ae1e-d7ca9417b453',
    }

data_search = requests.get(url_search, params=params_search)

data_search_json = data_search.json()['result']['results']
resource_id = data_search_json[0]['resources'][0]['id']
resource_format = data_search_json[0]['resources'][0]['format']
resource_id, resource_format
```

Coneción con datastore_search para obtener el dataset de package_search

```python
url_datastore = 'https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search'
params_datastore = {'resource_id':resource_id,
                    'limit':100}

response_datastore = requests.get(url_datastore, params_datastore)
response_datastore.json()
```

datastore devuelve help, success, result y records. records es el campo que ofrece los datos del dataset.

```python
datos_datastore = response_datastore.json()['result']['records']
datos_df = pd.DataFrame(datos_datastore)
```

```python
datos_df
```

### package_show

```python
# Coneción y parametros a la API con package_show
url_show = 'https://opendata-ajuntament.barcelona.cat/data/api/action/package_show'
params = {
    'id':'1412100d-b0cd-4894-ae1e-d7ca9417b453'
}

data_api_show = requests.get(url_show, params=params)
```

## 2 - De los resultados obtenidos, selecciona uno y exportalo a CSV o JSON

```python
# Obtengo los datos que necesito (resource id)
data_show = data_api_show.json()['result']
resource_id_show = data_show['resources'][0]['id']
```

## 3 - Con resource_id del recurso seleccionado, realiza una consulta con datastore_search para recuperar al menos 100 registros

Por defecto el limit ya es 100, pero se define igualmente.

```python
# Conexión y parametros a la API con datastore_search
url_datastore = 'https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search'
params_show_datastore = {'resource_id':resource_id_show,
                         'limit':100}

response_datastore = requests.get(url_datastore, params_show_datastore)

# Obtengo los datos del resource que quiero convertir a dataframe
responde_show_ds = response_datastore.json()
data = responde_show_ds['result']['records']
```

## 4 - Crea un DataFrame de los resultados

```python
# Convierto los datos al dataframe
show_df = pd.DataFrame(data)
show_df
```
