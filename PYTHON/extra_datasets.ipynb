```python
import pandas as pd
df = pd.read_csv("hospitals.csv", sep=";", encoding="latin-1")
df.head(5)
```

2. Realitza els exercicis de filtres següents:

» Mostra totes les dades dels hospitals que tenen entre 100 i 120 "num_camas".

» Mostra els noms dels hospitals que estan a Barcelona - Catalunya.

» Mostra els 3 hospitals amb més "indice_satisfaccion" del grup d'hospitals de Madrid i Andalusia.

» Mostra els hospitals que comencen per lletra "C" i terminen per lletra "a".

```python
df[(df["num_camas"]>=100)&(df["num_camas"]<=200)]
df[df["num_camas"].between(100,200,"both")]
df[(df["localidad"]=="Barcelona")]["nombre"]
df[(df["localidad"]=="Madrid")|(df["comunidad_autonoma"]=="Andalucia")].nlargest(3,"indice_satisfaccion").sort_values("indice_satisfaccion",ascending=False)
df[(df["localidad"]=="Madrid")].nlargest(3,"indice_satisfaccion").sort_values("indice_satisfaccion",ascending=False)
df[df["nombre"].str.startswith("Hospital C")&(df["nombre"].str.endswith("a"))]["nombre"]
```

3. Realitza els següents exercicis d'agrupacions:

» Mostra el "indice_satisfaccion" mig de cada comunitat autònoma i província de la taula "hospitals".

» Mostra el "num_camas" total de cada comunitat autònoma.

» Mostra el "porcentaje_ocupacion" més petit de cada província de cada comunitat autònoma.

» Mostra quantes províncies i localitats té cada comunitat autònoma.

» Mostra les comunitats autònomes que tenen menys de 5 hospitals.

```python
round(df.groupby(df["comunidad_autonoma"])["indice_satisfaccion"].mean(),2)
round(df.groupby(df["provincia"])["indice_satisfaccion"].mean(),2)
df.groupby(df["comunidad_autonoma"])["num_camas"].sum()
df.groupby(["comunidad_autonoma", "provincia"])["porcentaje_ocupacion"].min()
df.groupby(df["comunidad_autonoma"])[["provincia","localidad"]].nunique()
df.groupby(df["comunidad_autonoma"])["hospital_id"].nunique()
```
